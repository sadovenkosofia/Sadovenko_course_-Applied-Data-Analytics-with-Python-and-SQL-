```python
import numpy as np
import pandas as pd
```

```python
# Creating a DataFrame from raw data
data = {
    'first_name': ['Jason', 'Molly', 'Tina', 'Jake', 'Amy'],
    'last_name': ['Miller', 'Jacobson', '.', 'Milner', 'Cooze'],
    'age': [42, 52, 42, 24, 73],
    'preTestScore': [4, 24, 31, np.nan, np.nan],
    'postTestScore': [25 ,94 , 57, 62, 70]
}

df = pd.DataFrame(data)
print("DataFrame created:")
print(df)
```

```python
print(df.isnull())
```

```python
print(df.age.isnull())
```

```python
df.preTestScore = df.preTestScore.fillna(df.preTestScore.mean())
print(df)
```
