```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)
y = np.sin(x)
categories = ['A', 'B', 'C', 'D']
values = [3, 7, 2, 5]
data = np.random.randn(1000)

fig, axs = plt.subplots(2, 2, figsize=(12, 8))

axs[0, 0].plot(x, y, label='sin(x)', color='blue')
axs[0, 0].set_title('Sine Function')
axs[0, 0].legend()
axs[0, 0].grid(True, linestyle='--', alpha=0.6)

x_random = np.random.rand(50)
y_random = np.random.rand(50)
axs[0, 1].scatter(x_random, y_random, color='red')
axs[0, 1].set_title('Random Scatter Plot')
axs[0, 1].grid(True, linestyle='--', alpha=0.6)

axs[1, 0].bar(categories, values, color='purple')
axs[1, 0].set_title('Bar Plot Example')

axs[1, 1].hist(data, bins=30, color='green', alpha=0.7, edgecolor='black')
axs[1, 1].set_title('Histogram of Random Data')

plt.tight_layout()
plt.show()
```
