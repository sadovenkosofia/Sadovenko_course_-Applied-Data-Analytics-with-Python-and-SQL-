```python
import numpy as np
import pandas as pd

import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline

# Load a sample dataset
tips = sns.load_dataset("tips")

# Display the first few rows of the dataset
print(tips.head())
```

```python
# Scatter plot
sns.scatterplot(x='total_bill', y='tip', data=tips, hue='day')
plt.title('Scatter Plot of Tip vs Total Bill')
plt.show()
```

```python
# Scatter plot
sns.lineplot(x='total_bill', y='tip', data=tips)
plt.title('Line Plot of Tip vs Total Bill')
plt.show()
```

```python
# Histogram and KDE plot - shows distribution of a single continuous variable
sns.histplot(tips['total_bill'], kde=True)
plt.title('Distribution of Total Bill')
plt.show()
```

```python
# Box plot - Compare distributions across categories:
sns.boxplot(x='day', y='total_bill', data=tips)
plt.title('Box Plot of Total Bill by Day')
plt.show()
```

```python
# Pair plot - Visualize pairwise relationships in a dataset:
sns.pairplot(tips, hue='day')
plt.show()
```

```python
# Bar plot - Show the distribution of a categorical variable:
sns.barplot(x='day', y='total_bill', data=tips)
plt.title('Average Total Bill by Day')
plt.show()
```

```python
# Line Plot - used to display trends over time of continuous variables

# Scatter plot
sns.lineplot(x='total_bill', y='tip', data=tips)
plt.title("Line Plot of Tip vs Total Bill")
plt.show()
```

```python
tips_numerical = tips[['total_bill', 'tip']]
```

```python
# Heatmap
correlation_matrix = tips_numerical.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
