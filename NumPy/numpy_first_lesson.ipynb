```python
import numpy as np

#Python list example
python_list = [1, 2, 3, 4]
python_result = [x * 2 for x in python_list]

#Numpy list example
numpy_list = np.array([1, 2, 3, 4])
numpy_result = numpy_list * 2

print(python_result)
print(numpy_result)
```

```python
# 2D array
two_d_array = np.array([[1, 2, 3], [1, 2 ,3]])
print(two_d_array)
```

```python
#Example with addition

array = np.array([1, 2, 3])
array = array + 3
print(array)
```

```python
# Example with math function
arr = np.array([1, 2, 3, 4])
print(f'sum = {np.sum(arr)}')
print(f'mean = {np.mean(arr)}')
print(f'standart deviation = {np.std(arr)}')
```

```python
#Matrix multiplication
matrix_1 = np.array([[1 , 2], [2, 3]])
matrix_2 = np.array([[1, 2], [2, 3]])
result = np.dot(matrix_1, matrix_2)
print(result)
```
